```python
from digistudio.processing.upload import load_collection

cjs = categorized_jobs = load_collection("jobs-display")
display = cjs.iloc[1].to_dict()
```

```python
recent = cjs.iloc[0].to_dict()
len(recent['high'])
```

```python
def generate_enhanced_task(dictionary):
    sentences = []
    for data in dictionary: 
        metadata = data.get('metadata')
        companyInfo = metadata.get('company_info')
        general = metadata.get('general_info')
        company = companyInfo.get('hiringOrganization', '[Company Name]')
        position = general.get('employmentTitle', '[Position Title]')
        jobUrl = data.get('jobUrl')
        sentence = f"- {position} at {company} - please apply utilizing the following link as a starting point: [{jobUrl}]({jobUrl})"
        sentences.append(sentence)
    return sentences

open_positions = generate_enhanced_task(display['high'])
```

```python
num_of_pos = len(open_positions)
opening = f"Here are {num_of_pos} postions that I would like for your team to apply to today:"

closing = '''I would like for you team to:
- Add the job to my Huntr Job Board
- Update this sentence: "I am particularly drawn to [Company Name] because of its focus on [insert product area, mission, or technical challenge], and I see strong alignment between your needs and my background in AI-enabled products, data-informed decision-making, and delivery ownership."
- Submit Job Application'''

with open("daily_task_two.md", "w") as f:
    f.write(opening + "\n\n" + "\n\n".join(open_positions) + "\n\n" + closing)
```

```python
sample = "I am particularly drawn to [Company Name] because of its focus on [insert product area, mission, or technical challenge], and I see strong alignment between your needs and my background in AI-enabled products, data-informed decision-making, and delivery ownership."

data = display['high']

entries = []
for item in data: 
    info = item.get('metadata', {})
    ginfo = info.get('general_info', {})
    cinfo = info.get('company_info', {})
    sinfo = info.get('salary_info', {})
    stats = item.get('stats', {})
    position = ginfo.get('employmentTitle', '[Position Title]')
    company = cinfo.get('hiringOrganization', '[Company Name]')
    salary = sinfo.get('minValue', '[Salary Range]') + sinfo.get('maxValue', '[Salary Range]') / 2
    description = ginfo.get('description', '[Job Description]')
    score = stats.get('score', '[Relevance Score]')
    rationale = stats.get('rationale', '[Rationale for Score]')
    jobUrl = item.get('jobUrl', '[Job URL]')

    entry = {
        "Position": position,
        "Company": company,
        "Salary": f"${salary:,.2f}",
        "Description": description,
        "Score": score,
        "Rationale": rationale,
        "Url": jobUrl
    }
    entries.append(entry)
```

```python
import pandas as pd

pd.DataFrame(entries).to_csv("jobs.csv")
```
