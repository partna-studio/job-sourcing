# LinkedIn Job Scraper API Test

This notebook tests the Node.js Express server running on `http://localhost:3000`

## Import Libraries

```python
import requests
import json
from pprint import pprint
```

## Configuration

```python
BASE_URL = 'http://localhost:3000'
HEALTH_ENDPOINT = f'{BASE_URL}/api/health'
JOBS_ENDPOINT = f'{BASE_URL}/api/jobs'

print(f'Testing API at: {BASE_URL}')
```

## Test 1: Health Check

```python
try:
    response = requests.get(HEALTH_ENDPOINT, timeout=5)
    print(f'Status Code: {response.status_code}')
    print(f'Response: {response.json()}')
except requests.exceptions.ConnectionError:
    print('❌ Error: Cannot connect to server. Is npm start running?')
except Exception as e:
    print(f'❌ Error: {e}')
```

## Test 2: Fetch LinkedIn Jobs

```python
try:
    print('Fetching jobs from LinkedIn...')
    response = requests.get(JOBS_ENDPOINT, timeout=30)
    response.raise_for_status()
    
    data = response.json()
    print(f'\n✅ Success!')
    print(f'Status: {data.get("success")}')
    print(f'Jobs Found: {data.get("count")}')
    
except requests.exceptions.Timeout:
    print('❌ Error: Request timed out. LinkedIn API may be slow.')
except requests.exceptions.ConnectionError:
    print('❌ Error: Cannot connect to server. Is npm start running?')
except requests.exceptions.HTTPError as e:
    print(f'❌ HTTP Error: {e}')
except Exception as e:
    print(f'❌ Error: {e}')
```

## Test 3: Display Jobs Data

```python
if data.get('success'):
    jobs = data.get('data', [])
    print(f'\nTotal Jobs: {len(jobs)}\n')
    
    for i, job in enumerate(jobs[:3], 1):
        print(f'--- Job {i} ---')
        print(f'Title: {job.get("title", "N/A")}')
        print(f'Company: {job.get("company", "N/A")}')
        print(f'Location: {job.get("location", "N/A")}')
        print()
```

## Test 5: Full Response JSON

```python
print('Full Response (first job only):')
if data.get('success') and len(data.get('data', [])) > 0:
    pprint(data['data'][0])
else:
    print('No jobs available')
```

## Test 6: Send Custom Query Options (POST Request)

```python
try:
    # Define custom query options
    custom_options = {
        'keyword': 'python developer',
        'location': 'United States',
        'dateSincePosted': 'past Month',
        'jobType': 'full time',
        'remoteFilter': 'remote',
        'salary': '150000',
        'experienceLevel': 'mid level',
        'limit': '15',
        'page': '0',
        'has_verification': False,
        'under_10_applicants': False,
    }
    
    print('Sending custom query options...')
    print(f'Options: {custom_options}\n')
    
    # Send POST request with custom options
    response = requests.post(JOBS_ENDPOINT, json=custom_options, timeout=30)
    response.raise_for_status()
    
    data = response.json()
    print(f'✅ Success!')
    print(f'Jobs Found: {data.get("count")}')\n    print(f'Query options used: {data.get("queryOptions")}')\n    
except Exception as e:
    print(f'❌ Error: {e}')
```
